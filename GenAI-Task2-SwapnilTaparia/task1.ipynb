```python
import sys
```

```python
TAX_AMOUNT = 0.05

DISCOUNT_AMOUNT_DICT = {
    2000: 0.15,
    1500: 0.1,
    1000: 0.05
}

try:
    order_amount = float(input('Enter the order amount'))
except ValueError:
    print('Enter a valid input')
    sys.exit()

for (amount, discount_level) in DISCOUNT_AMOUNT_DICT.items():
    if order_amount >= amount:
        discount_percentage = f"{discount_level * 100}%"
        discounted_price = order_amount * (1 - discount_level)
        tax_amount = TAX_AMOUNT * discounted_price
        break
    


print(f"Order Value: {order_amount}\n")
print(f"Subtotal (after {discount_percentage} off): {discounted_price}")
print(f"Tax: {tax_amount}")
print(f"Total: {order_amount + tax_amount}")
```
