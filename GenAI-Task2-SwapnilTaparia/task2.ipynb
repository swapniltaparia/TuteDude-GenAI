```python
DISCOUNT_AMOUNT_DICT = {
    2000: 0.15,
    1500: 0.1,
    1000: 0.05
}

orders = [1200, 2500, 800, 1750, 3000]

def getTotals(order_amount):
    discount_percentage = 0
    discounted_price = order_amount

    for (amount, discount_level) in DISCOUNT_AMOUNT_DICT.items():
        if order_amount >= amount:
            discount_percentage = discount_level * 100
            discounted_price = order_amount * (1 - discount_level)
            break
    
    return (
        order_amount,
        discount_percentage,
        discounted_price,
        discounted_price,
    )

total_revenue = 0

products_info = [getTotals(order_amount) for order_amount in orders]

for (
    order_amount,
    discount_percent,
    discount_price,
    total
) in products_info:
    print(f"{order_amount} -> {f'{discount_percent}% off'} -> {total}")

    total_revenue += total

print()
print(f"Total Products: {len(products_info)}")
print(f"Total Discounted Products: {sum(product[1] > 0 for product in products_info)}")

print()
print(f"Total Revenue (after discounts): {total_revenue}")
```
