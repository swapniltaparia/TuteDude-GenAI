```python
menu_options = {'1','2','q'}

while (True):
    user_input = input("""
        Menu options:
            1. Add order amount
            2.Show Order
            q: Quit
    """).strip()

    if user_input in menu_options and user_input == 'q':
        break
    else:
        if user_input not in menu_options:
            print('Invalid option')
        continue
```
