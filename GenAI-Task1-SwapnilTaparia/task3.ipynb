```python
price_dict = {
    'Wireless Bluetooth Headphones': 2499,
    'Stainless Steel Insulated Water Bottle': 799,
    'Premium Cotton Oversized T-Shirt': 899,
    'Minimalist Leather Wallet': 1199,
    'Smart LED Desk Lamp': 999,
    'Portable Mini Bluetooth Speaker': 599,
}
```

```python
# Change price of existing product
price_dict['Ceramic Coffee Mug'] = 399

print(price_dict)

# Existing Key
key_1 = 'Wireless Bluetooth Headphones'

# Non-existing key
key_2 = 'Ergonomic Wireless Mouse'

del price_dict[key_1]

# handling Key Error
# Method 1
try:
    del price_dict[key_2]
except Exception as e:
    print(f"{type(e).__name__}: {e}")

# Method 2
if key_2 in price_dict:
    del price_dict[key_2]

# Method 3
price_dict.pop(key_2, None)


print(price_dict)
```
