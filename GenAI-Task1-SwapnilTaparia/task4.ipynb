```python
products = [
    'Wireless Bluetooth Headphones',
    'Stainless Steel Insulated Water Bottle',
    'Premium Cotton Oversized T-Shirt',
    'Minimalist Leather Wallet',
    'Smart LED Desk Lamp',
    'Portable Mini Bluetooth Speaker'
]

price_dict = {
    'Wireless Bluetooth Headphones': 2499,
    'Stainless Steel Insulated Water Bottle': 799,
    'Premium Cotton Oversized T-Shirt': 899,
    'Minimalist Leather Wallet': 1199,
    'Smart LED Desk Lamp': 999,
    'Portable Mini Bluetooth Speaker': 599,
}

categories = [
    'Electronics',
    'Kitchen',
    'Apparels',
    'Accessories',
    'Home Decor',
    'Electronics'
]
```

```python
product_info = [
    (product, price_dict[product], categories[index])
    for index, product in enumerate(products)
]
```

```python
category_to_products = {}

for name, price, category in product_info:
    if category not in category_to_products:
        category_to_products[category] = []
    
    category_to_products[category].append(name)

print(category_to_products)
```

```python
max_products_category = max(category_to_products, key=lambda x: len(category_to_products[x]))

print(category_to_products[max_products_category])
```
