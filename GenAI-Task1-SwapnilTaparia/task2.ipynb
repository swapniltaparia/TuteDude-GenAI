```python
categories = [
    'Electronics',
    'Kitchen',
    'Apparels',
    'Accessories',
    'Home Decor',
    'Electronics'
]

categories_set = set(categories)

print(categories_set)
# Because 'Electronics' category is duplicate, the categories length is 5 while products length is 6
```

```python
# Add a new category
categories_set.add('Groceries')



length_original_set = len(categories_set)

categories_set.add('Home Decor')
length_modified_set = len(categories_set)

# checking if length before and after adding a duplicate item is same or not
print(length_original_set == length_modified_set)

print(categories_set, length_modified_set)
```
