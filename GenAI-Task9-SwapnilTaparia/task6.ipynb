```python
import numpy as np
```

```python
marks=np.array([78, 85, 90, 66, 72, 88, 95, 60])
```

```python
print(np.sort(marks))
print(np.percentile(marks, 25))
print(np.percentile(marks, 50))
print(np.percentile(marks, 75))
print(np.sort(marks))
```

```python
average = np.mean(marks)

print(np.sum(marks > average))
```
