```python
import numpy as np
```

```python
# 1D array of integers from 1 to 10
array_1d=np.arange(1, 11)

# 2D array of shape (3,3) with integers from 1 to 9
array_2d=np.arange(1, 10).reshape(3,3)

# A Numpy array from list [10, 20, 30, 40, 50]
array_from_list=np.array([10,20,30,40,50])

print(array_2d)
print(array_1d)
print(array_from_list)
```

```python
print(array_1d.shape, array_1d.dtype)
print(array_2d.shape, array_2d.dtype)
print(array_from_list.shape, array_from_list.dtype)
```
