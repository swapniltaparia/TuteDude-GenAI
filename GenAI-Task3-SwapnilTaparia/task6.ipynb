```python
def process_prices(prices):
    discounted_prices = list(map(lambda price: price * 0.9, prices))
    filtered_prices = list(filter(lambda discounted_price: discounted_price > 300 , discounted_prices))

    return (
        discounted_prices,
        filtered_prices
    )

process_prices([100, 500, 900, 50, 750])
```
