```python
def add_price(prices_list, price):
    prices_list.append(price)

def get_average_price(prices_list):
    if len(prices_list) == 0:
        return 0

    return sum(prices_list) / len(prices_list)

def get_max_price(prices_list):
    return max(prices_list, default = 0)

VALID_MENU_OPTIONS = ['1', '2', '3', 'q']

prices_list = []

while True:
    user_input = input("""
        Enter input:
            1 -> Add price
            2 -> Show average price
            3 -> Show highest price

            q -> Quit                
    """)

    if user_input not in VALID_MENU_OPTIONS:
        print(f"You have not entered `{user_input}` which is not a valid option\n\n")
        continue

    if user_input == 'q':
        print(f"Price list: {prices_list}")
        print(f"Average of the price list: {get_average_price(prices_list)}")
        print(f"Highest price of the list: {get_max_price(prices_list)}\n\n")

        break

    if user_input == '1':
        while True:
            try:
                user_price_input = int(input("Enter price to add"))
            except ValueError:
                print('You entered an invalid price\n\n')
                continue

            if user_price_input <= 0:
                print('You entered a non-positive price\n\n')
                continue
                
            print(f"Original price list: {prices_list}")
            add_price(prices_list, user_price_input)
            print(f"Price list with {user_price_input} added: {prices_list}\n")
            break

    if user_input == '2':
        print(f"Average of the price list: {get_average_price(prices_list)}\n")

    if user_input == '3':
        print(f"Highest price of the list: {get_max_price(prices_list)}\n")


```
